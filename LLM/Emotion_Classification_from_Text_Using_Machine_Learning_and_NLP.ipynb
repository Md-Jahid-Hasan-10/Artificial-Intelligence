```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep = ';',header = None,names = ['text','emotion'])
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotion'].unique()
unique_emotions
```

##oputput converts numbers

```python
emotion_numbers = {}
i = 0
for emo in unique_emotions:
  emotion_numbers[emo] = i
  i +=1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

## convert Lowercase Letter

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

## remove punctuation

```python
import string

def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

## remove numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
import emoji
from bs4 import BeautifulSoup
import re
```

```python
def remove_html_tags(text):
    if not isinstance(text, str): return ""
    return BeautifulSoup(text, "html.parser").get_text()
df['text'] = df['text'].apply(remove_html_tags)
```

```python
def remove_urls(text):
    return re.sub(r'https?://\S+|www\.\S+|ftp://\S+', '', text)
df['text'] = df['text'].apply(remove_urls)
```

```python
def remove_emojicons(text):
    return emoji.replace_emoji(text, replace='')
df['text'] = df['text'].apply(remove_emojicons)
```

# Tokenization and Remove Stop words

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
def remove(txt):
  words = word_tokenize(txt)
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

# Using Bag of Words and Naive Bayes algo

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)


nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)


pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
```

#Using TFIDF and Naive Bayes algo

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)


nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf,y_train)
```

```python
y_pred = nb2_model.predict(X_test_tfidf)
print(accuracy_score(y_test, y_pred))
```
