```python
df = pd.read_csv('/content/ML/anime.csv')
df.head()
```

```python
df.loc[0]['Title']
```

```python
def ex_episode(txt):
  data = ""
  ok = False
  for i in txt:
    if i == ')':
      return data
    if ok:
      data += i
    if i == '(':
      ok = True
  return data
```

```python
df["Episode"] = df["Title"].apply(ex_episode)
```
